```python
import h5py
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import loguniform, gaussian_kde
from pycbc import conversions
```

### Load the population model ($\mathcal{M}_{c}$ distribution) and the FF results from the precessing study

```python
mf = h5py.File('sensitivity/pop-models/nsbh/trani/mchirp.hdf')
FF_f = h5py.File('/work/rahul.dhurkunde/searches/banksim/ecc-injections-precess-spin-new/30000_FFs.hdf')
```

```python
FF = FF_f['FF']
FF_mchirp = conversions.mchirp_from_mass1_mass2(FF_f['mass1'][:], FF_f['mass2'][:])
```

### Get a KDE of the mchirp samples $p(\mathcal{M}_c)$

```python
mchirp = mf['samples'][:]
kde = gaussian_kde(mchirp)
```

### Obtain the average bias in merger rate over the parameter space

$bias$ = 1 - $\dfrac{\sum FF^3 \times p(\mathcal{M}_c)}{\sum p(\mathcal{M}_c)}$

```python
avg_FF = np.sum(np.power(FF[:10000],3) * kde(FF_mchirp[:10000])) /  np.sum(kde(FF_mchirp[:10000]))
print('Average bias:', (1-avg_FF)*100, '%')
```
